```python
from collections import Counter

def load_dictionary(file_path):
    """Load the Scrabble dictionary from a file."""
    with open(file_path, 'r') as file:
        # Assuming the file has one word per line
        return [word.strip().lower() for word in file.readlines()]

def can_form_word(word, letters):
    """Check if a word can be formed using the given letters."""
    word_counter = Counter(word)
    letters_counter = Counter(letters)
    return all(word_counter[char] <= letters_counter[char] for char in word_counter)

def find_valid_words(dictionary, letters):
    """Find all words in the dictionary that can be formed with the given letters."""
    return [word for word in dictionary if can_form_word(word, letters)]

def main():
    # Define the letters and dictionary file path
    letters = "tabind"
    dictionary_file = "Collins_Scrabble_Words_2019.txt"  # Update this path if needed

    # Load the dictionary
    dictionary = load_dictionary(dictionary_file)

    # Find valid words
    valid_words = find_valid_words(dictionary, letters)

    # Sort the words alphabetically
    valid_words.sort()

    # Output the result
    print(f"Valid words formed from '{letters}':")
    for word in valid_words:
        print(word)

    # Print the total number of valid words
    print(f"\nTotal number of valid words: {len(valid_words)}")

if __name__ == "__main__":
    main()
```
